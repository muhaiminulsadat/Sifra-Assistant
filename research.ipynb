```python
import os
from google import genai
from dotenv import load_dotenv

load_dotenv(override=True)

def gemini_model_response(user_input):
    # The SDK automatically looks for 'GOOGLE_API_KEY' in environment variables.
    # If it's not set, you can pass it directly:
    client = genai.Client(api_key=os.getenv('GEMINI_API_KEY'))
    
    # In 2025, 'gemini-2.0-flash' is the stable high-speed model.
    # For reasoning, use 'gemini-2.0-pro-preview'.
    model_id = "gemini-2.0-flash" 
    
    prompt = f"Give me the sentiment of this sentence: {user_input}"
    
    # New syntax: client.models.generate_content
    response = client.models.generate_content(
        model=model_id,
        contents=prompt
    )
    
    return response.text

# Example test
# print(gemini_model_response("I am so happy this finally works!"))
```

```python
gemini_model_response('How are you?')
```

```python
import os
from groq import Groq


def groq_model_response(prompt_text):
    client = Groq(api_key=os.getenv("GROQ_API_KEY"))
    
    # Using the latest versatile model for 2025
    model_id = "llama-3.3-70b-versatile"

    chat_completion = client.chat.completions.create(
        # We wrap your string in the required 'user' role dictionary here
        messages=[
            {"role": "user", "content": prompt_text}
        ],
        model=model_id,
        temperature=0.5,
    )

    return chat_completion.choices[0].message.content

# Now you can use it exactly how you wanted:
# response = groq_model_response("How are you today?")
# print(response)
```

```python
groq_model_response('Hi')
```

```python
from openai import OpenAI
def groq_openai_response(prompt: str, model: str = "openai/gpt-oss-20b") -> str:

    # Initialize client
    client = OpenAI(
        api_key=os.getenv("GROQ_API_KEY"),
        base_url="https://api.groq.com/openai/v1"
    )

    # Call the model
    response = client.responses.create(
        model=model,
        input=prompt
    )

    return response.output_text
```

```python
groq_openai_response('HI')
```
